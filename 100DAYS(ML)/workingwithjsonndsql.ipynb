```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("students.db")

with open("students.sql", "r") as file:
    sql = file.read()

conn.executescript(sql)

df = pd.read_sql("SELECT * FROM students", conn)

print(df)
```
