```python
# here we have only 2 category but still if we have many category then we can use the most frequent one only 
counts=flights['time'].value_counts()
replace=counts[counts<=100].index
```

```python
new_time = flights['time'].astype(str).replace(replace, 'others')

pd.get_dummies(new_time, dtype=int).sample(10)
```
