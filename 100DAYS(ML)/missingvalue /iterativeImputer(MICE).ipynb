```python
import pandas as pd
import numpy as np

# 5-row dataset
df = pd.DataFrame({
    'Age':    [22, np.nan, 35, 28, np.nan],
    'Fare':   [7.25, 71.28, np.nan, 8.05, 31.50],
    'SibSp':  [1, 1, 0, np.nan, 2]
})

df
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

mice = IterativeImputer(random_state=42)

df_mice = mice.fit_transform(df)

df_mice = pd.DataFrame(
    df_mice,
    columns=df.columns
)

df_mice
```
