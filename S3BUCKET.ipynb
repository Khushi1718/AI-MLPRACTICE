```python
#download csv file
import boto3
import pandas as pd
```

```python
s3=boto3.resource(
    service='s3',
    region_name='us-east-2',
    aws_acsess_key_id='',
    aws_secret_access_key=''

)
#IAM = we add user , access type = program access,
```

```python
#print out bucket name
for bucket in s3.buckets.add():
    print(bucket.name)
```
