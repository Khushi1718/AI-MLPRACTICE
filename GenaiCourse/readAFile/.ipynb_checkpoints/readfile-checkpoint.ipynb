```python
# read a whole file
with open('example.txt','r') as file :
    content = file.read()
    print(content)
```

```python
# read a file line by line 
with open ('example.txt','r')as file:
    for line in file :
        # print(line)
        print(line.strip()) # strip() removes the newline charcter
```

```python
#writting a file(overwritting)
with open('example.txt' ,'w') as file:
    file.write('Hello i am good\n')
    file.write('i am leaning file management today')

```

```python
# write in a file without overwritting:
with open('example.txt','a') as file:
    file.write('\nappening the text \n')
    file.write('one more')
```

```python
# writing a list of lines in a file 
lines=['file first \n','second line \n','third line']
with open('example.txt','a')as file:
    file.writelines(lines)
```
