```python
# Binary files - we generally update bits 

#writitng to a binary file
data = b'\x00\x01\x02\x03'
with open('example.bin','wb') as file:
    file.write(data)

# reading a binary file
with open ('example.bin','rb')as file:
    content = file.read()
    print(content)



```

```python
# read the text file and count the number of lines ,words and characters
def count_text_file(file):
    with open(file,'r') as f:
        lines = f.readlines()
        count_line = len(lines)
        count_word=0
        count_char=0
        for line in lines :
            for word in line:
                count_word = count_word+1
                for char in word:
                    count_char= count_char+1
    return count_line, count_word,count_char
line,word,char = count_text_file('example.txt')
print(f'Lines: {line} , Words:{word} , Char:{char}')
```
